Индивидуальное задание

```python
class Stack:
    def __init__(self):   # Используем список для хранения элементов
        self.items = []

    def push(self, item):    # Добавляет элемент на вершину стека
        self.items.append(item)

    def pop(self):    # Удаляет и возвращает элемент с вершины стека
        return self.items.pop()

    def peek(self):  # Возвращает верхний элемент, не удаляя его
        return self.items[-1]

    def is_empty(self):  # Проверяет, пуст ли стек
        return len(self.items) == 0
```

```python
# Идея: стек, как и скобки, имеет структуру Last In First Out, значит, мы можем класть каждую открывающую скобку в стек, и при закрытии удалять ее из стека
def is_balanced(expression):
    stack = Stack()
    match_brackets = {')': '(', ']': '[', '}': '{'}  # Сопоставим скобки с помощью словаря
    # Рассмотрим скобочки и кавычки отдельно: скобкам нужен другой элемент для закрытия. кавычкам - такой же
    for char in expression:
        if char in '([{':
            stack.push(char)
        elif char in ')]}':
            if stack.is_empty() or stack.pop() != match_brackets[char]:
                return False
                
        elif char in '"\'':
            if not stack.is_empty() and stack.peek() == char:
                stack.pop()
            else:    # Иначе это открывающая кавычка, кладем в стек
                stack.push(char)
                
    # Если после обработки всей строки стек пуст - все скобки сбалансированы
    return stack.is_empty()
```

```python
text0 = "jabbbjhvs(jbv)kjabv{}jsn[jbsv(kjnv{jnv})]"
text1 = "({[]}'')"
text2 = 'snbn"nbv[jv]nvs(nvb{bnv})"'   # Первые 3 без ошибок
text3 = "jbvbv)"
text4 = 'ksnv"hvb"nvb(njv'
text5 = 'fvbsuub(jbvs{iusbv[bs])}nfv'
text = [text0, text1, text2, text3, text4, text5]
for _ in text:
    print(is_balanced(_))
```
