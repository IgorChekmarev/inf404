```python
# Код
from collections import deque    # Добавим двустороннюю очередь. Идейно тот же список, но в обычном списке изменение на позиции 0 оценивается сложностью O(n), а в deque - сложностью O(1). Так что можно и без этого, но зачем?

def is_bipartite(graph):   # Делаем функцию, которая через словарь будет помечать точки (закрашенная/незакрашенная). Идем через BFS (поиск в ширину)
    colors = {}
    
    for start_dot in graph:    # Проходим по всем вершинам, чтобы найти несвязные графы
        if start_dot not in colors:    # Начинаем обход BFS для новой компоненты связности
            colors[start_dot] = 0
            queue = deque([start_dot])
            
            while queue:
                u = queue.popleft()    # Вот здесь как раз работает deque - вынимаем 0 элемент и удаляем его из списка. В list'е это заняло бы больше времени
                
                for v in graph[u]:
                    if v not in colors:    # Красим соседа в противоположный цвет
                        colors[v] = 1 - colors[u]
                        queue.append(v)
                    elif colors[v] == colors[u]:
                        return False, None      # Если сосед уже покрашен в тот же цвет - это конфликт, граф не является двудольным (есть нечетный цикл).
                        
    return True, colors     # Если конфликтов не было, граф двудольный
```

```python
# Тесты

print("Двудольный граф (квадрат/четный цикл)")
graph_bipartite = {0: [1, 3],   # Сам граф
                   1: [0, 2],
                   2: [1, 3],
                   3: [0, 2]}
    
is_bi, colors = is_bipartite(graph_bipartite)   # По написанной функции, получаем True или False на двудольноть и словарь на цвета
print(f"Двудольность {is_bi}")
if is_bi:
    print(f"Раскраска: {colors}")     # Еси двудольный - раскраска есть. Выводим
else:
    print("Раскраска невозможна, граф не двудольный")     # Если не двудольный - в colors будет None. Пусть будет что-то поподробнее на выводе в тесте
print()     # Строчка, чисто для разделения тестов

# Остальные тесты идейно такие же, но на другие графы. Не вижу смысла их комментировать

######################################################################


print("Граф с нечетным циклом (треугольник)")
graph_odd_cycle = {0: [1, 2],
                   1: [0, 2],
                   2: [0, 1]}

is_bi, colors = is_bipartite(graph_odd_cycle)
print(f"Двудольность {is_bi}")
if is_bi:
    print(f"Раскраска: {colors}")
else:
    print("Раскраска невозможна, граф не двудольный")
print()


########################################################################


print("Несвязный двудольный граф")
graph_disconnected = {0: [1],
                      1: [0],
                      2: [3],
                      3: [2]}

is_bip, colors = is_bipartite(graph_disconnected)
print(f"Двудольность {is_bip}")
if is_bip:
    print(f"Раскраска: {colors}")
else:
    print("Раскраска невозможна, граф не двудольный")
```
